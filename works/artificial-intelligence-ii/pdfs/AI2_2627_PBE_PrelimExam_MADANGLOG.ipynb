```python
#Part A.1
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
#Load the dataset and programmatically determine:
df = pd.read_csv('smart_home_energy.csv')

df.info()
#number of observations
print(f'observation: \n',df.shape[0])
#number of predictor variables
print(f'predictors: \n',df.shape[1]-1)
#data types
print(f'datatype: \n',df.dtypes)
#missing values
print(f'Missing Values: \n',df.isnull().sum())
#duplicate observations
print(f'duplicate observation: \n',df.duplicated().sum())
#descriptive statistics
print(f'statistics: \n',df.describe)

```

```python
#Part A.2. You must identify the dependent and independent variables yourself.
#Produce at least three appropriate visualizations that help you understand relationships among the variables.
#At least one visualization must investigate relationships between predictors and Energy_Consumption.

#HeatMap
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot = True , fmt = '.2f', cmap = "coolwarm")
plt.title("correlation heatmap")
plt.tight_layout()
plt.show()

#scatter plot
plt.figure(figsize=(6, 4))
sns.scatterplot(x='AC_Hours', y='Energy_Consumption', data=df)
plt.title("AC_Hours vs Energy_Consumption")
plt.show()

#histogram
plt.figure(figsize=(6, 4))
sns.histplot(df['Energy_Consumption'], kde= True)
plt.title("Energy_Consumption")
plt.show()
```
