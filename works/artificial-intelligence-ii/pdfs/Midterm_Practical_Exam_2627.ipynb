#1. Tokenization & Cleaning Task

Given your assigned documents, create a preprocessing function that:

1. Tokenizes the text.
2. Converts tokens to lowercase.
3. Removes tokens containing only punctuation.
4. Keeps alphabetic words.
5. Removes standalone numbers.
6. Prints the tokens containing numbers separately.
7. Prints the final cleaned tokens.
8. Count the number of tokens for both original and cleaned tokens.

```python
#YOUR CODES FOR TASKS #1 GOES HERE. MAKE SURE TO ADD COMMENTS DESCRIBING BLOCKS OF CODES FOR EACH NUMBER

import nltk
import string
import re
from nltk.tokenize import word_tokenize

documents = [
    "The company's revenue increased from $125,000 to $148,500.",
    "Operating costs were approximately $52,750.25.",
    "The annual growth rate reached 18.4%.",
    "The company expects revenue to exceed $200,000 next year."
]

def preprocess_text(text):
  tokens = re.findall(r"\b[A-Za-z]+|\d+(?:,\d{3})*(?:\.\d+)?", text)

  tokens = [token.lower() for token in tokens]

  original_count = len(tokens)

  number_tokens = [
      token for token in tokens
      if re.search(r"\d", token)
  ]

  print("tokens Containing Numbers")
  print(number_tokens)

  cleaned_tokens = [
      token for token in tokens
      if re.fullmatch(r"[a-z,A-Z]+", token)
  ]

  print("\nCleaned Tokens")
  print(cleaned_tokens)

  print("\nOriginal Token Count")
  print("Cleaned Token Count", len(cleaned_tokens))

  return cleaned_tokens

cleaned_documents =[]


for i, document in enumerate(documents, start = 1):
  print("_____________")
  print("\nDOCUMENT", i)


  cleaned = preprocess_text(document)
  cleaned_documents.append(cleaned)
```

##2. Stemming, Lemmatization and POS-tagging

Using the same documents assigned, write the codes for the following:

1. Apply stemming - PorterStemmer
2. Apply lemmatization
3. Identify POS-tags

```python
##YOUR CODES FOR TASKS #2 GOES HERE. MAKE SURE TO ADD COMMENTS DESCRIBING BLOCKS OF CODES FOR EACH NUMBER
import re
import nltk

from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag


stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

for i, tokens in enumerate(cleaned_documents,start=1):
  print("_____________")
  print("\nDOCUMENT", i)

  words = [
      word for word in tokens
      if re.fullmatch(r"[a-z,A-Z]+", word)
  ]

  stemmed_words = [
      stemmer.stem(word)
      for word in words
  ]

  print("\nSTEMMED WORDS")
  print(stemmed_words)

  lemmatized_words = [
      lemmatizer.lemmatize(word)
      for word in words
  ]

  print("\nLemmatized Words")
  print(lemmatized_words)

  pos_words = pos_tag(words)

  print("\nPOS TAGS")
  print(pos_words)
```
